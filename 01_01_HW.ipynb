Напишите программу, проверяющую стаж работы сотрудника. В переменной worker содержится список, 
содержащий значения как во 2-ом задании (имя, фамилия, зарплата, стаж). 
Поместите в переменную status такую строку

user_name user_family is position

где user_name и user_family - имя и фамилия сотрудника, а position - одна из следующих строк

* 'junior' - стаж меньше 2 лет
* 'middle' - стаж больше либо равен 2 лет
* 'senior' - стаж больше 5 лет 

Пример данных, которые могут быть в worker

worker = ['Olya', 'Silyutina', 350000, 3]
Hint: печатать результат не нужно, просто задайте его в переменную status

```python
worker = ['Marina', 'Vystavkina', 28000, 4]
experience = worker[-1]

if experience < 2:
    position = 'junior'
elif 2 <= experience <= 5:
    position = 'middle'
else:
    position = 'senior'

# status = '{user_name} {user_family} is {position}'.format(user_name = worker[0], user_family = worker[1], position = position)
status = '{} {} is\033[1m {} \033[0mstaff.'
status = status.format(worker[0], worker[1], position)
print(status)
status
```

```python
values = [12, 134, 10, 47, 100, 33, 20, 50, 160, 210, 111]
tens = []
no_tens = []
for value in values:
    if value % 10 == 0:
        tens.append(value)
    else:
        no_tens.append(value)

print('Divisible by 10 are - ', tens)
print('Not divisible by 10 are - ', no_tens)
```

```python
workers = [['Ivan', 'Ivanov', 100000, 20], ['Petr', 'Petrov', 150000, 1], ['Sidor', 'Sidorov', 200000, 3]]

for worker in workers:
    experience = worker[-1]
    if experience < 2:
        position = 'junior'
    elif 2 <= experience <= 5:
        position = 'middle'
    else:
        position = 'senior'
    status = '{} {} is {} staff.'
    status = status.format(worker[0], worker[1], position)
    print(status)
```
