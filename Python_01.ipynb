```python
# 1. Добавление элементов:
# append() - добавляет элемент в конец списка
fruits = ['apple', 'banana']
fruits.append('orange')  # ['apple', 'banana', 'orange']

# insert() - добавляет элемент в указанную позицию в списке
fruits.insert(1, 'mango')  # ['apple', 'mango', 'banana', 'orange']

# extend() - добавляет несколько элементов из другого списка
more_fruits = ['grape', 'kiwi']
fruits.extend(more_fruits)  # ['apple', 'mango', 'banana', 'orange', 'grape', 'kiwi']

# 2. Removing Elements:
# remove() - removes first occurrence of value
fruits.remove('banana')  # removes 'banana'

# pop() - removes and returns element at index (default: last)
last_fruit = fruits.pop()  # removes and returns last element
specific_fruit = fruits.pop(1)  # removes and returns element at index 1

# clear() - removes all elements
fruits.clear()  # []

# 3. Finding Elements:
fruits = ['apple', 'banana', 'orange', 'banana']

# index() - returns first index of value
banana_index = fruits.index('banana')  # returns 1

# count() - counts occurrences of value
banana_count = fruits.count('banana')  # returns 2

# 4. Ordering:
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

# sort() - sorts list in-place
numbers.sort()  # [1, 1, 2, 3, 4, 5, 6, 9]
numbers.sort(reverse=True)  # [9, 6, 5, 4, 3, 2, 1, 1]

# reverse() - reverses list in-place
numbers.reverse()  # reverses current order

# 5. Copying:
# copy() - returns shallow copy of list
original = [1, 2, 3]
new_list = original.copy()

# 6. Additional Operations:
# len() - gets length of list (not a method, but common function)
length = len(fruits)

# in - checks if element exists (not a method, but common operator)
has_apple = 'apple' in fruits  # True/False
```

```python
job_years = 5
```

```python
salary = 100000
```

```python
message_old = 'Bonus is equal to '
```

```python
# {} - означает, что строку можно форматировать = подставлять что нужно вместо '{}'
message = 'Bonus is equal to {coef}%. New salary is {new_salary} roubles'
```

```python
if job_years < 1:
    coef = 0
elif job_years >= 1 and job_years <= 2:
    coef = 0.2
else:
    coef = 0.4

new_salary = salary + salary * coef

# складываем = конкатенируем строки
print(message_old + str(coef*100) + '%')
print(new_salary, 'roubles')
```

```python
if job_years < 1:
    coef = 0
elif job_years >= 1 and job_years <= 2:
    coef = 0.2
else:
    coef = 0.4

new_salary = salary + salary * coef

# применяем метод .format() к переменной 'message'
# т.е. содержимое () метода .format подставится в строку на место {}
print(message.format(coef=coef*100, new_salary=new_salary))
```

```python
analysts = ['Anatoly', 'Olga', 'Beslan']
```

```python
# цикл FOR-IN:
for i in analysts:
    print(i)
    print("Analyst's name is {}".format(i))
```

```python
# FOR + IF → двойной отступ для содержимого IF-clause 
for i in analysts:
    if i == 'Anatoly':
        print("Analyst's name is {}".format(i))
```

```python
# словари - задаются фигурными {} скобками
# словари - это пары "Ключ:Значение"
# в качестве ЗНАЧЕНИЯ может выступать список
analysts_data = {'Anatoly': [2, 100000], 'Olga': [3, 200000], 'Beslan': [4, 300000]}
```

```python
analysts_data
```

```python
# достаем первый элемент с "полки" 'Anatoly'
analysts_data['Anatoly'][1]
```

```python
# добавим новый элемент в словарь
analysts_data['Michael'] = [10, 400000]
```

```python
analysts_data
```

```python
# список доступен по НАЗВАНИЯМ КЛЮЧЕЙ - НЕ по ИНДЕКСУ
analysts_data[0]
```

```python
for i in analysts_data:
    print(i) # только КЛЮЧ
    print(analysts_data[i]) # только ЗНАЧЕНИЕ
    print(i, analysts_data[i]) # и КЛЮЧ и ЗНАЧЕНИЕ
    print("Analyst's name is {}.".format(i))
    print("Salary is {}.".format(analysts_data[i][1]))
    print("Work experience is {} years.".format(analysts_data[i][0]))
```

```python
for i in analysts_data:
    if analysts_data[i][0] > 3: # доступ к элементу ключа → Имя_Списка[Имя_Элемента][Номер_Позиции_в_Ключе]
        print(i)
        print('Work experience - {} years.'.format(analysts_data[i][0]))
```
