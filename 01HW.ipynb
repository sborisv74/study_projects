```python
workers = [['Ivan', 'Ivanov', 100000, 1], ['Petr', 'Petrov', 150000, 8], ['Sidor', 'Sidorov', 200000, 3]]
```

```python
for worker in workers:
    if worker[3] < 2: # или worker[-1]; experience = worker[3]
        position = 'Junior'
    elif worker[3] > 5:
        position = 'Senior'
    else:
        position = 'Middle'
    status = '{} {} is {}'.format(worker[0], worker[1], position)
    print(status)
```
