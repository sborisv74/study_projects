```python
# Basic Arithmetic Operators

# Addition (+): Adds two numbers together.
# Example: 5 + 2 = 7
print(5 + 2)

# Subtraction (-): Subtracts one number from another.
# Example: 4 - 2 = 2
print(4 - 2)

# Multiplication (*): Multiplies two numbers.
# Example: 2 * 3 = 6
print(2 * 3)

# Division (/): Divides one number by another, returning a float result.
# Example: 4 / 2 = 2.0
print(4 / 2)

# Floor Division (//): Divides and returns the integer value by discarding the fractional part.
# Example: 9 // 2 = 4
print(9 // 2)

# Modulus (%): Returns the remainder of dividing one number by another.
# Example: 5 % 2 = 1
print(5 % 2)

# Exponentiation (**): Raises one number to the power of another.
# Example: 4 ** 2 = 16
print(4 ** 2)
```
